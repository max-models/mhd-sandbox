# Tutorials

The notebooks in this section run the solver on the classic ideal MHD test problems. They are
executed when the documentation is built; download a notebook from its page to run it yourself.

- [Orszag-Tang vortex](/mhd-sandbox/tutorials/orszag-tang/): the standard 2D test, from smooth
  vortices to interacting shocks, as a snapshot and an animation.
- [Ideal MHD test problems](/mhd-sandbox/tutorials/test-problems/): the Orszag-Tang vortex, the
  MHD rotor and a magnetized blast wave, with their energies and profiles.
- [Multi-fluid MHD](/mhd-sandbox/tutorials/multifluid/): several ideal MHD fluids advanced
  together with one time step.
